## A student is buying a laptop.

    Laptop price: Rs. 85,000
    Student has: Rs. 100,000
    Student gets a discount of Rs. 5,000
    A laptop bag costs Rs. 2,500
    The store has a promotion code "SAVE10".

### Create variables and:

    Calculate the laptop price after discount.
    Add the bag price to get the final bill.
    Calculate the remaining money.
    Check whether the student has enough money.
    Check whether "SAVE10" is present in the promotion code.
    Check whether the final bill is greater than Rs. 80,000.
    Store the result of whether the student has enough money in a variable and check whether it is truthy.

```python
# Variables
laptop_price = 85000
student_money = 100000
discount = 5000
bag_price = 2500
promo_code = "SAVE10"

# Calculations
price_after_discount = laptop_price - discount
final_bill = price_after_discount + bag_price
remaining_money = student_money - final_bill

# Logical checks
has_enough_money = student_money >= final_bill
is_promo_present = "SAVE10" in promo_code
bill_greater_than_80k = final_bill > 80000

# Checking truthy status of our budget check variable
has_enough_money_truthy = bool(has_enough_money)

# Results
print("1. Laptop price after discount: Rs.", price_after_discount)
print("2. Final bill (with bag): Rs.", final_bill)
print("3. Remaining money: Rs.", remaining_money)
print("4. Has enough money:", has_enough_money)
print("5. 'SAVE10' present in promo code:", is_promo_present)
print("6. Final bill greater than Rs. 80,000:", bill_greater_than_80k)
print("7. Is 'has_enough_money' status truthy?:", has_enough_money_truthy)
```

### You are creating a student result calculator.

A student has:

    Mathematics: 85
    English: 72
    Python: 90
    Total possible marks per subject: 100
    Student name: "Ahmed"
    Required passing marks per subject: 50

### Create variables and:

    Calculate the student's total marks.
    Calculate the average marks.
    Calculate the percentage.
    Check whether the student passed Mathematics.
    Check whether the student passed all three subjects.
    Check whether the name contains the letter "A".
    Check whether the average marks are greater than or equal to 80.
    Store the final percentage and check whether it is truthy.

Print all results.

```python
# Variables
math = 85
english = 72
python = 90
total_possible_per_subject = 100
student_name = "Ahmed"
passing_marks = 50

# Calculations
total_marks = math + english + python
average_marks = total_marks / 3
percentage = (total_marks / (total_possible_per_subject * 3)) * 100

# Condition Checks
passed_math = math >= passing_marks
passed_all = (math >= passing_marks) and (english >= passing_marks) and (python >= passing_marks)
name_has_A = "A" in student_name
avg_greater_or_equal_80 = average_marks >= 80
percentage_truthy = bool(percentage)

# Printing all results
print(f"--- Result Card for {student_name} ---")
print("Total Marks Scored:", total_marks)
print("Average Marks:", round(average_marks, 2))
print("Percentage:", round(percentage, 2), "%")
print("Passed Mathematics:", passed_math)
print("Passed All Three Subjects:", passed_all)
print("Does name contain letter 'A'?:", name_has_A)
print("Average marks >= 80:", avg_greater_or_equal_80)
print("Is the final percentage value truthy?:", percentage_truthy)
```
